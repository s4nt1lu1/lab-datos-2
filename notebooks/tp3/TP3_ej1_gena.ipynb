# Ejercicio 1:

Diccionario de variables — Dataset Banco Mundial

Claves de unión

- **country**: Nombre del país.
- **year**: Año al que corresponde la observación.

Índices de libertad económica — Heritage Foundation

- **heritage_score**: Puntaje global del Index of Economic Freedom. Resume el desempeño en 12 componentes de libertad económica; valores más altos indican mayor libertad económica.
- **property_rights**: Protección de los derechos de propiedad, incluyendo la capacidad de personas y empresas de acumular, utilizar y transferir propiedad privada de manera segura.
- **government_integrity**: Integridad del gobierno y percepción de prácticas como corrupción, sobornos, nepotismo y captura institucional. Valores más altos = mayor integridad.
- **judicial_effectiveness**: Eficacia, independencia y calidad del sistema judicial y su capacidad de hacer cumplir leyes y contratos.
- **tax_burden**: Libertad fiscal, construida a partir de la carga impositiva sobre individuos y empresas y de la recaudación tributaria. Valores más altos = menor carga fiscal relativa.
- **government_spending**: Nivel de gasto público y peso del gobierno en la economía (es un índice de Heritage, no el gasto público directo como % del PIB).
- **fiscal_health**: Situación fiscal del gobierno, considerando principalmente déficit presupuestario y deuda pública.
- **business_freedom**: Facilidad para iniciar, operar y cerrar una empresa, según el entorno regulatorio y administrativo.
- **labor_freedom**: Flexibilidad y eficiencia del mercado laboral (contratación, salarios, despidos).
- **monetary_freedom**: Estabilidad de precios y grado de intervención/control de precios. Valores más altos = mayor estabilidad monetaria.
- **trade_freedom**: Ausencia de barreras arancelarias y no arancelarias al comercio internacional.
- **investment_freedom**: Libertad para mover y utilizar capital de inversión, nacional y extranjero, con pocas restricciones regulatorias.
- **financial_freedom**: Independencia y eficiencia del sistema financiero y grado de intervención gubernamental en bancos y servicios financieros.
- **iso_code**: Código ISO de tres letras del país, usado por Heritage Foundation.

Índices de libertad económica — Fraser Institute (EFW)

- **efw_score**: Puntaje global del índice Economic Freedom of the World. Resume cinco grandes áreas; valores más altos = mayor libertad económica.
- **size_of_government**: Área 1 del EFW — tamaño del gobierno (gasto público, impuestos, empresas estatales, transferencias).
- **legal_system_and_property_rights**: Área 2 del EFW — calidad del sistema legal, independencia judicial, cumplimiento de contratos y protección de la propiedad.
- **sound_money**: Área 3 del EFW — estabilidad monetaria, inflación y acceso a una moneda relativamente estable.
- **freedom_to_trade_internationally**: Área 4 del EFW — libertad para comerciar internacionalmente (aranceles, barreras regulatorias, controles de capitales y divisas).
- **regulation_of_credit_labor_and_business**: Área 5 del EFW — grado de regulación de los mercados de crédito, trabajo y actividad empresarial.
- **standard_deviation_of_the_5_efw_areas**: Desviación estándar entre las cinco áreas del EFW. Valores mayores = mayor heterogeneidad entre dimensiones de libertad económica.
- **countryiso3**: Código ISO 3166-1 alfa-3 del país, usado por el Banco Mundial.

## Inciso 2:

La idea de este análisis es comparar la mediana de gdp_per_capita_current_us entre los países que sí reportan cada una de las variables identificadas con muchos faltantes y los que no la reportan. La diferencia porcentual entre ambos grupos permite ver si la ausencia del dato está asociada a países más pobres o más ricos (es decir, si el faltante es informativo y no aleatorio). Por ejemplo, podemos deducir que los países más pobres tienden a no reportar el valor de high-technology_exports_of_manufactured_exports.

Por lo tanto podemos decir que las ausencias de datos es informativo (MNAR) y no aleatorio

```python
target = 'gdp_per_capita_current_us$'

for col in variables_no_validas:
    con_dato = df_2.loc[df_2[col].notna(), target]
    sin_dato = df_2.loc[df_2[col].isna(), target]
    mediana_con = con_dato.median()
    mediana_sin = sin_dato.median()
    diferencia_pct = (mediana_con - mediana_sin) / mediana_sin * 100
    print(f"{col}:")
    print(f"  Mediana PIB per cápita CON dato: {mediana_con:.0f}")
    print(f"  Mediana PIB per cápita SIN dato: {mediana_sin:.0f}")
    print(f"  Diferencia: {diferencia_pct:.1f}%\n")
```

```python
# Todas las columnas con al menos un faltante
cols_con_faltantes = df_2.columns[df_2.isna().any()].tolist()

resumen_faltantes = []
for col in cols_con_faltantes:
    if col == target:
        continue
    pct_faltante = df_2[col].isna().mean() * 100
    con_dato = df_2.loc[df_2[col].notna(), target].median()
    sin_dato = df_2.loc[df_2[col].isna(), target].median()
    diferencia_pct = (con_dato - sin_dato) / sin_dato * 100
    resumen_faltantes.append({
        'variable': col,
        'pct_faltante': round(pct_faltante, 1),
        'mediana_con_dato': round(con_dato, 0),
        'mediana_sin_dato': round(sin_dato, 0),
        'diferencia_pct': round(diferencia_pct, 1)
    })

resumen_faltantes = pd.DataFrame(resumen_faltantes).sort_values('pct_faltante', ascending=False)
print(resumen_faltantes.to_string(index=False))
```

Mayor a 50% de faltantes (adult_literacy_rate, fiscal_health, judicial_effectiveness, central_government_debt_of_gdp, gini_index, poverty x2): acá imputar sería en su mayoría inventar datos (75-80% del dato sería "adivinado"). Creamos la bandera _is_missing y descartamos el valor original.

10-50% de faltantes (high-technology_exports, real_interest_rate, school_enrollment_tertiary/secondary/primary, labor_freedom, government_expenditure_on_education, domestic_credit_to_private_sector, energy_use, freedom_to_trade_internationally): todavía hay suficiente dato real como para que imputar aporte algo. Creamos la bandera _is_missing e imputamos con la mediana.

<10% de faltantes (el resto): el impacto de imputar es marginal por la poca cantidad de filas afectadas. Imputamos directo con la mediana, sin necesidad de bandera.

```python
from sklearn.impute import KNNImputer
from sklearn.preprocessing import StandardScaler

target = 'gdp_per_capita_current_us$'

# 0. Filas sin el target: no se pueden usar ni para entrenar ni para evaluar
df_4 = df_2.dropna(subset=[target]).copy()
print(f"Filas descartadas por falta de target: {df_2.shape[0] - df_4.shape[0]}")

# 1. % de faltantes de cada variable numérica (sin contar el target)
numericas_4 = df_4.select_dtypes(include=[np.number]).columns.tolist()
numericas_4.remove(target)
pct_faltante = df_4[numericas_4].isna().mean() * 100

# 2. Armamos los tres grupos según el umbral
grupo_alto = pct_faltante[pct_faltante > 50].index.tolist()
grupo_medio = pct_faltante[(pct_faltante > 10) & (pct_faltante <= 50)].index.tolist()
grupo_bajo = pct_faltante[(pct_faltante > 0) & (pct_faltante <= 10)].index.tolist()

print(f"Grupo alto (>50%, se descarta el valor original): {grupo_alto}")
print(f"\nGrupo medio (10-50%, bandera + imputación): {grupo_medio}")
print(f"\nGrupo bajo (<10%, imputación directa): {grupo_bajo}")

# 3. GRUPO ALTO: creamos la bandera de faltante y descartamos el valor original
for col in grupo_alto:
    df_4[f'{col}_is_missing'] = df_4[col].isna().astype(int)
df_4 = df_4.drop(columns=grupo_alto)

# GRUPO MEDIO: bandera de faltante + imputación con KNN (en vez de mediana)
for col in grupo_medio:
    df_4[f'{col}_is_missing'] = df_4[col].isna().astype(int)

scaler_temp = StandardScaler()
X_medio_scaled = scaler_temp.fit_transform(df_4[grupo_medio])

knn_imputer = KNNImputer(n_neighbors=5)
X_medio_imputado_scaled = knn_imputer.fit_transform(X_medio_scaled)

# Volvemos a la escala original de cada variable
df_4[grupo_medio] = scaler_temp.inverse_transform(X_medio_imputado_scaled)

print("Faltantes restantes en grupo medio:", df_4[grupo_medio].isna().sum().sum())

# 5. GRUPO BAJO: imputación directa con la mediana, sin bandera
for col in grupo_bajo:
    df_4[col] = df_4[col].fillna(df_4[col].median())

print(f"\nShape final: {df_4.shape}")
print(f"Faltantes restantes: {df_4.isna().sum().sum()}")
```

```python
# Excluimos las banderas de faltante y 'year': no son candidatas a transformación de asimetría
# Recalculamos las columnas numéricas actuales de df_4 (después del tratamiento de faltantes)
numericas_4 = df_4.select_dtypes(include=[np.number]).columns.tolist()
numericas_4.remove(target)

# Excluimos las banderas de faltante y 'year': no son candidatas a transformación de asimetría
cols_is_missing = [c for c in numericas_4 if c.endswith('_is_missing')]
candidatas = [c for c in numericas_4 if c not in cols_is_missing and c != 'year']

skewness_filtrada = df_4[candidatas].skew().sort_values(key=abs, ascending=False)
print(skewness_filtrada.head(15))

# Mínimo de cada una de las top 10 más asimétricas, para saber qué transformación es válida
top_asimetricas = skewness_filtrada.head(10).index.tolist()
for col in top_asimetricas:
    print(f"{col}: mínimo = {df_4[col].min():.2f}, skew = {df_4[col].skew():.2f}")
```

```python
from scipy import stats

data_num = df_4[candidatas].copy()

# Logaritmo natural (log1p para evitar log(0))
data_log = np.log1p(data_num)

# Raíz cuadrada
data_sqrt = np.sqrt(data_num)

# Box-Cox: requiere que los datos sean estrictamente positivos
resultados_boxcox = {}
for col in data_num.columns:
    if (data_num[col] > 0).all():
        transformado, lam = stats.boxcox(data_num[col])
        resultados_boxcox[col] = {
            'lambda_boxcox': lam,
            'skew_boxcox': pd.Series(transformado).skew()
        }
    else:
        resultados_boxcox[col] = {'lambda_boxcox': None, 'skew_boxcox': None}

# Yeo-Johnson: puede manejar ceros y negativos
resultados_yeojohnson = {}
for col in data_num.columns:
    transformado, lam = stats.yeojohnson(data_num[col])
    resultados_yeojohnson[col] = {
        'lambda_yeojohnson': lam,
        'skew_yeojohnson': pd.Series(transformado).skew()
    }

```

```python
display(pd.DataFrame({
    'Original': data_num.skew(),
    'Logaritmo Natural': data_log.skew(),
    'Raíz Cuadrada': data_sqrt.skew(),
    'Box-Cox': [resultados_boxcox[col]['skew_boxcox'] for col in data_num.columns if col in resultados_boxcox],
    'Yeo-Johnson': [resultados_yeojohnson[col]['skew_yeojohnson'] for col in data_num.columns if col in resultados_yeojohnson]
}).round(3))
```

Se puede observar que para las variables numericas que toman valores positivos, los metodos de Yeo-Jhonson y Box-Cox dan valores similares e incluso iguales, esto es porque Yeo-Jhonson es la extension de Box-Cox para valores negativos y nulos, para valores positivos realizan el mismo metodo y tiene sentido que den lo mismo. En general casi para todas las variables el Metodo de Yeo-Jhonson realiza mejores transformaciones que la transforacion logaritmica y de raiz, log y raíz son transformaciones "fijas" (siempre aplican la misma fórmula), mientras que Box-Cox y Yeo-Johnson buscan automáticamente el mejor parámetro λ (lambda) que minimiza la asimetría para cada variable en particular por eso se ajustan mucho mejor, variable por variable. Aunque se pueden observar variables donde Yeo-Jhonson empeora el resultado de la asimetria, como por ejemplo gdp_growth_annual y real_interest_rate, esto pasa porque estas variables tienen su asimetría dominada por unos pocos valores extremos (una hiperinflación puntual, una crisis de desinversión en un país-año específico), y ninguna transformación de potencia puede "arreglar" eso del todo

Podemos ver que hay variables donde Log le gana a Yeo, veamos el porque:

```python
print("NaN generados por log1p:")
print(data_log[['inflation_consumer_prices_annual', 'gdp_growth_annual', 'gdp_per_capita_growth_annual']].isna().sum())


print("\nNaN generados por raíz cuadrada:")
print(data_sqrt[['inflation_consumer_prices_annual', 'gdp_growth_annual', 'gdp_per_capita_growth_annual']].isna().sum())
```

Podemos observar que para las variables donde le ganan a Yeo no se esta haciendo la comparación justa, ya que ambas trasformaciones no estan definidas para valores negativos y generaron valores faltantes en estas variables

```python
data_yeojohnson_final = pd.DataFrame(index=data_num.index)

for col in data_num.columns:
    transformado, lam = stats.yeojohnson(data_num[col])
    data_yeojohnson_final[col] = transformado
```

Es necesario llevar las variables a una escala común cuando la magnitud de una de ellas es excesivamente mas grande que las otras por ejemplo gdp_current_us$ en miles de millones vs heritage_score entre 0 y 100. Ademas es necesario la estandarizacion para modelos que son sensibles a la distribuciones y escala de los datos como pueden ser KNN o SVM, al contrario de metodos que son robustos a las distribuciones sesgadas y outliers como Arboles de decision, Random Forest porque dividen por umbrales de orden, por lo tanto escalar no sirve de nada

```python
from sklearn.preprocessing import StandardScaler

scaler = StandardScaler()
data_escalada = pd.DataFrame(
    scaler.fit_transform(data_yeojohnson_final),
    columns=data_yeojohnson_final.columns,
    index=data_yeojohnson_final.index
)

print("Media antes de escalar (2 variables de ejemplo):")
print(data_yeojohnson_final[['gdp_current_us$', 'heritage_score']].mean())

print("\nMedia después de escalar (debería ser ~0):")
print(data_escalada[['gdp_current_us$', 'heritage_score']].mean().round(6))
print("\nDesvío después de escalar (debería ser ~1):")
print(data_escalada[['gdp_current_us$', 'heritage_score']].std().round(6))
```
